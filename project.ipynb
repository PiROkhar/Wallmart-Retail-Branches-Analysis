Data Exploration and loading

```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('Walmart.csv', encoding_errors='ignore')
df.shape
```

```python
df.info()
```

```python
#Total Duplicate
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df['unit_price'] = df['unit_price'].str.replace('$','').astype(float)
```

```python
df.head()
```

```python
df['quantity'] = df['quantity'].astype(int)
df.head()
```

```python
df['total'] = df['unit_price'] * df['quantity']
df.head()
```

```python
# mysql
# host = localhost
# port = 3306
# user = root
# password = 1920
```

```python
#Connecting to mysql

engine_sql = create_engine("mysql+pymysql://root:1920@localhost:3306/wallmart")

try:
    engine_sql
    print("Connection Successful")
except:
    print("Connection failed")
```

```python
df.to_sql(name = 'wallmart', con = engine_sql, if_exists='append', index=False)
```

```python
df.columns = df.columns.str.lower()
df['total'] = df['total'].round(2)
df.head()
```

```python
df.to_csv('cleaned_dataset', index = False)
```
